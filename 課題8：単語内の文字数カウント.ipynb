```python
word_list = [] #入力された英単語を格納する空リスト
word_chr = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"),0) #参照するためのアルファベットの辞書
```

```python
while True : #入力が空でない間繰り返す  
    eng_word = input("英単語を入力してください：")
    lower_eng_word = eng_word.lower()
    print(f"英単語を入力してください：{lower_eng_word}")

    if lower_eng_word != "": #入力が空でない場合
        sp_eng_word = list(lower_eng_word) #入力された英単語を一文字ずつに分解
        word_list.append(lower_eng_word) #入力された英単語を辞書word_listに格納
        for i in sp_eng_word:
            word_chr[i] += 1 #word_chrの辞書に登場した数を加算
            
    else:
        word_list.sort()
        print(f"入力した英単語：{word_list}")

        break
```

```python
for key, value in word_chr.items():
    if value >= 1:    
        print(f"{key}が{value}個ありました")
    else:
        pass
```
